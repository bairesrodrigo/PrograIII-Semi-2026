```python
class AnalizadorMultiuso:
    """Clase con utilidades para procesamiento matemático, estadístico y de cadenas."""

    @staticmethod
    def calcular_media_armonica(numeros):
        """Calcula la media armónica de una lista de números positivos."""
        if not numeros:
            return 0.0
        # Validar que no existan valores <= 0 (la media armónica no acepta ceros o negativos)
        if any(x <= 0 for x in numeros):
            raise ValueError("Todos los valores deben ser estrictamente mayores que 0.")

        n = len(numeros)
        suma_reciprocos = sum(1.0 / x for x in numeros)
        return n / suma_reciprocos

    @staticmethod
    def calcular_frecuencias(datos):
        """Cuenta la cantidad de apariciones de cada elemento (caracteres o palabras)."""
        frecuencias = {}
        for elemento in datos:
            frecuencias[elemento] = frecuencias.get(elemento, 0) + 1
        return frecuencias

    @staticmethod
    def es_palindromo(texto):
        """Verifica si un texto se lee igual de izquierda a derecha y viceversa."""
        # Limpieza: eliminar espacios y convertir a minúsculas
        texto_limpio = "".join(c.lower() for c in texto if c.isalnum())
        if not texto_limpio:
            return False
        return texto_limpio == texto_limpio[::-1]


def menu_ejercicio():
    analizador = AnalizadorMultiuso()

    while True:
        print("\n" + "=" * 50)
        print("    EJERCICIO: MEDIA ARMÓNICA, FRECUENCIA Y PALÍNDROMO")
        print("=" * 50)
        print("  [1] Calcular Media Armónica")
        print("  [2] Calcular Frecuencia de caracteres o palabras")
        print("  [3] Verificar si un texto es Palíndromo")
        print("  [4] Salir")
        print("=" * 50)

        opcion = input("Seleccione una opción (1-4): ").strip()

        if opcion == "1":
            print("\n--- MEDIA ARMÓNICA ---")
            entrada = input("Ingrese números positivos separados por coma (ej: 4, 8, 12): ")
            try:
                numeros = [float(x.strip()) for x in entrada.split(",") if x.strip()]
                resultado = analizador.calcular_media_armonica(numeros)
                print(f"Datos ingresados : {numeros}")
                print(f"Media Armónica   : {round(resultado, 4)}")
            except ValueError as e:
                print(f"[!] Error de entrada: {e}")

        elif opcion == "2":
            print("\n--- FRECUENCIA DE ELEMENTOS ---")
            tipo = input("¿Desea contar [C]aracteres o [P]alabras?: ").strip().lower()
            texto = input("Ingrese el texto a analizar: ").strip()

            if tipo == "p":
                datos = texto.split()
            else:
                datos = list(texto)

            frecuencias = analizador.calcular_frecuencias(datos)
            print("\nConteo de frecuencias:")
            for clave, valor in frecuencias.items():
                print(f"  • '{clave}': {valor} vez/veces")

        elif opcion == "3":
            print("\n--- COMPROBADOR DE PALÍNDROMOS ---")
            cadena = input("Ingrese una palabra o frase: ").strip()
            es_pal = analizador.es_palindromo(cadena)

            if es_pal:
                print(f"✓ La frase/palabra '{cadena}' SÍ es un palíndromo.")
            else:
                print(f"✗ La frase/palabra '{cadena}' NO es un palíndromo.")

        elif opcion == "4":
            print("\n¡Ejecución finalizada!")
            break
        else:
            print("\n[!] Opción inválida. Seleccione un número del 1 al 4.")


if __name__ == "__main__":
    menu_ejercicio()
```

```python
import json
import math

"""
===============================================================================
SISTEMA ACADÉMICO Y ESTADÍSTICO INTERACTIVO - UGB
===============================================================================
"""

class SistemaEstadistico:
    """Clase encargada del procesamiento y cálculo estadístico de una lista numérica."""
    def __init__(self, datos, nombre_variable="Variable"):
        self.datos = datos
        self.nombre_variable = nombre_variable
        self.n = len(datos)

    def calcular_media(self):
        if self.n == 0: return 0
        return sum(self.datos) / self.n

    def calcular_mediana(self):
        if self.n == 0: return 0
        datos_ordenados = sorted(self.datos)
        mitad = self.n // 2
        if self.n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2.0
        else:
            return datos_ordenados[mitad]

    def calcular_frecuencias(self):
        frecuencias = {}
        for dato in self.datos:
            frecuencias[dato] = frecuencias.get(dato, 0) + 1
        return frecuencias

    def calcular_moda(self):
        if self.n == 0: return []
        frecuencias = self.calcular_frecuencias()
        max_freq = max(frecuencias.values())
        return [clave for clave, valor in frecuencias.items() if valor == max_freq]

    def calcular_varianza(self, es_muestra=True):
        if self.n <= 1: return 0
        media = self.calcular_media()
        suma_cuadrados = sum((x - media) ** 2 for x in self.datos)
        divisor = (self.n - 1) if es_muestra else self.n
        return suma_cuadrados / divisor

    def calcular_desviacion_estandar(self, es_muestra=True):
        return math.sqrt(self.calcular_varianza(es_muestra))

    def calcular_rango(self):
        if self.n == 0: return 0
        return max(self.datos) - min(self.datos)

    def mostrar_reporte(self):
        """Imprime directamente la tabla con el reporte estadístico en consola."""
        print(f"\n" + "=" * 55)
        print(f" ANÁLISIS ESTADÍSTICO: {self.nombre_variable.upper()}")
        print("=" * 55)
        print(f"  • Total Registros (N)         : {self.n}")
        print(f"  • Media Aritmética            : {round(self.calcular_media(), 2)}")
        print(f"  • Mediana                     : {round(self.calcular_mediana(), 2)}")
        print(f"  • Moda                        : {self.calcular_moda()}")
        print(f"  • Varianza (Muestral)         : {round(self.calcular_varianza(), 2)}")
        print(f"  • Desviación Estándar / Típica: {round(self.calcular_desviacion_estandar(), 2)}")
        print(f"  • Rango                       : {round(self.calcular_rango(), 2)}")
        print(f"  • Frecuencias Absolutas       : {self.calcular_frecuencias()}")
        print("=" * 55)


class SistemaAcademico:
    """Clase para gestionar la base de datos de los estudiantes."""
    def __init__(self):
        self.estudiantes = []

    def cargar_datos_json(self, json_raw):
        self.estudiantes = json.loads(json_raw)

    def agregar_estudiante(self, id_est, nombre, materia, nota, edad, traslado, telefono):
        nuevo = {
            "id": id_est,
            "nombre": nombre,
            "materia": materia,
            "nota": float(nota),
            "edad": int(edad),
            "traslado_min": int(traslado),
            "uso_telefono_hrs": float(telefono)
        }
        self.estudiantes.append(nuevo)
        print(f"\n¡Estudiante '{nombre}' registrado exitosamente!")

    def extraer_variable(self, clave):
        return [e[clave] for e in self.estudiantes if clave in e]

    def exportar_json_formateado(self):
        return json.dumps(self.estudiantes, indent=4, ensure_ascii=False)


# FUNCIÓN DEL MENÚ INTERACTIVO
def ejecutar_menu_principal():
    # Dataset inicial de estudiantes
    json_dataset = '''[
        {"id": "001", "nombre": "Samuel Portillo", "materia": "Estadística Computacional", "nota": 10.0, "edad": 20, "traslado_min": 80, "uso_telefono_hrs": 6.0},
        {"id": "002", "nombre": "Carlos Turcios", "materia": "Admin. Bases de Datos II", "nota": 9.5, "edad": 20, "traslado_min": 25, "uso_telefono_hrs": 5.0},
        {"id": "003", "nombre": "Gilda González", "materia": "Circuitos Digitales", "nota": 9.3, "edad": 19, "traslado_min": 30, "uso_telefono_hrs": 5.9},
        {"id": "004", "nombre": "Saúl Jimenez", "materia": "Programación Comp. II", "nota": 10.0, "edad": 21, "traslado_min": 60, "uso_telefono_hrs": 5.0},
        {"id": "005", "nombre": "Edgar Ramirez", "materia": "Inglés Básico", "nota": 10.0, "edad": 21, "traslado_min": 120, "uso_telefono_hrs": 7.0},
        {"id": "006", "nombre": "Yeimy Parada", "materia": "Admin. Bases de Datos II", "nota": 9.5, "edad": 20, "traslado_min": 45, "uso_telefono_hrs": 3.0},
        {"id": "007", "nombre": "Miguel Gómez", "materia": "Programación Comp. III", "nota": 9.5, "edad": 19, "traslado_min": 15, "uso_telefono_hrs": 4.0},
        {"id": "008", "nombre": "Josué Funes", "materia": "Estadística Computacional", "nota": 9.5, "edad": 19, "traslado_min": 30, "uso_telefono_hrs": 5.0},
        {"id": "009", "nombre": "Jefferson Castillo", "materia": "Circuitos Digitales", "nota": 9.5, "edad": 19, "traslado_min": 15, "uso_telefono_hrs": 2.3},
        {"id": "010", "nombre": "Ramael Cordova", "materia": "Inglés Básico", "nota": 9.5, "edad": 22, "traslado_min": 7, "uso_telefono_hrs": 13.5},
        {"id": "011", "nombre": "Carlos Parada", "materia": "Programación Comp. III", "nota": 8.5, "edad": 25, "traslado_min": 10, "uso_telefono_hrs": 5.5},
        {"id": "012", "nombre": "Madison Valle", "materia": "Estadística Computacional", "nota": 10.0, "edad": 20, "traslado_min": 30, "uso_telefono_hrs": 9.0},
        {"id": "013", "nombre": "Jennifer Renderos", "materia": "Admin. Bases de Datos II", "nota": 9.6, "edad": 21, "traslado_min": 60, "uso_telefono_hrs": 3.0},
        {"id": "014", "nombre": "Diego Villafranco", "materia": "Circuitos Digitales", "nota": 9.0, "edad": 20, "traslado_min": 30, "uso_telefono_hrs": 8.0},
        {"id": "015", "nombre": "Flor Campos", "materia": "Estadística Computacional", "nota": 9.0, "edad": 20, "traslado_min": 6, "uso_telefono_hrs": 5.4},
        {"id": "016", "nombre": "Wilfredo Ortez", "materia": "Programación Comp. III", "nota": 9.5, "edad": 20, "traslado_min": 30, "uso_telefono_hrs": 5.0},
        {"id": "017", "nombre": "Kimberly Castro", "materia": "Programación Comp. III", "nota": 10.0, "edad": 18, "traslado_min": 20, "uso_telefono_hrs": 9.0}
    ]'''

    sistema = SistemaAcademico()
    sistema.cargar_datos_json(json_dataset)

    variables_map = {
        "1": ("nota", "Notas Académicas"),
        "2": ("edad", "Edades"),
        "3": ("traslado_min", "Tiempo de Traslado (Minutos)"),
        "4": ("uso_telefono_hrs", "Uso de Teléfono (Horas/Día)")
    }

    while True:
        print("\n" + "=" * 55)
        print("        MENÚ SISTEMA ACADÉMICO Y ESTADÍSTICO UGB")
        print("=" * 55)
        print("  [1] Ver dataset completo (Formato JSON)")
        print("  [2] Calcular estadísticas de UNA variable específica")
        print("  [3] Calcular estadísticas de TODAS las variables")
        print("  [4] Registrar un nuevo estudiante")
        print("  [5] Salir del programa")
        print("=" * 55)

        opcion = input("Seleccione una opción (1-5): ").strip()

        if opcion == "1":
            print("\n--- BASE DE DATOS EN FORMATO JSON ---")
            print(sistema.exportar_json_formateado())

        elif opcion == "2":
            print("\n--- SELECCIONE LA VARIABLE A ANALIZAR ---")
            print("  1. Notas Académicas")
            print("  2. Edades")
            print("  3. Tiempo de Traslado a la UGB (Minutos)")
            print("  4. Uso de Teléfono (Horas/Día)")

            sub_opcion = input("Elija una variable (1-4): ").strip()
            if sub_opcion in variables_map:
                clave, nombre = variables_map[sub_opcion]
                datos = sistema.extraer_variable(clave)
                analizador = SistemaEstadistico(datos, nombre_variable=nombre)
                analizador.mostrar_reporte()
            else:
                print("\n[!] Opción de variable no válida.")

        elif opcion == "3":
            for clave, nombre in variables_map.values():
                datos = sistema.extraer_variable(clave)
                analizador = SistemaEstadistico(datos, nombre_variable=nombre)
                analizador.mostrar_reporte()

        elif opcion == "4":
            print("\n--- INGRESO DE NUEVO ESTUDIANTE ---")
            try:
                id_est = input("ID (ej. 018): ").strip()
                nombre = input("Nombre completo: ").strip()
                materia = input("Materia: ").strip()
                nota = float(input("Nota (0.0 - 10.0): "))
                edad = int(input("Edad: "))
                traslado = int(input("Tiempo de traslado en minutos: "))
                telefono = float(input("Horas de uso de teléfono al día: "))

                sistema.agregar_estudiante(id_est, nombre, materia, nota, edad, traslado, telefono)
            except ValueError:
                print("\n[!] Error: Asegúrese de ingresar valores numéricos válidos en nota, edad, traslado y teléfono.")

        elif opcion == "5":
            print("\nSaliendo del sistema... ¡Gracias!")
            break

        else:
            print("\n[!] Opción inválida. Intente de nuevo con un número del 1 al 5.")


# Ejecución del programa
if __name__ == "__main__":
    ejecutar_menu_principal()
```
