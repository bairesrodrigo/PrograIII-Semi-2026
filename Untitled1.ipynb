```python
from functools import reduce

# 1. FUNCIÓN MAP
# Transformación elemento a elemento

# A. Extracción de iniciales de días de la semana
semana = ["domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado"]
print(f"Semana: {semana} | Primer día: {semana[0]}")

primera_letra = list(map(lambda dia: dia[0], semana))
print("Primeras letras:", primera_letra)

primera_letra_mayus = list(map(lambda dia: dia[0].upper(), semana))
print("Primeras letras (mayúsculas):", primera_letra_mayus)

# B. Elevar lista a un exponente dinamico
numeros_base = [2, 4, 6, 7, 9]
exponente = int(input("Exponente: "))
potencias = list(map(lambda num: num ** exponente, numeros_base))
print("Resultados potencia:", potencias)

# C. Generar tabla de multiplicar
numeros_tabla = list(range(1, 11))
tabla = int(input("Tabla a multiplicar: "))
resultado_tabla = list(map(lambda num: num * tabla, numeros_tabla))
print(f"Tabla del {tabla}:", resultado_tabla)


# 2. FUNCIÓN FILTER
# Filtrado condicional de elemento

# A. Clasificación de números pares e impares
serie = [2, 5, 6, 8, 9, 10, 12]
pares = list(filter(lambda num: num % 2 == 0, serie))
impares = list(filter(lambda num: num % 2 != 0, serie))

print("Original:", serie)
print("Pares:", pares)
print("Impares:", impares)

# B. Clasificación de notas (Lista estática)
notas = [1, 6, 8, 2, 7, 10, 5]
NOTA_MINIMA = 6

aprobados = list(filter(lambda nota: nota >= NOTA_MINIMA, notas))
reprobados = list(filter(lambda nota: nota < NOTA_MINIMA, notas))

print("Notas:", notas)
print("Aprobados:", aprobados)
print("Reprobados:", reprobados)

# C. Clasificación de notas desde input de usuario
entrada_usuario = input("Notas separadas por coma (ej: 10,8.8,6.5,4.5,3): ")
lista_notas = entrada_usuario.split(",")

aprobados_user = list(filter(lambda nota: float(nota) >= NOTA_MINIMA, lista_notas))
reprobados_user = list(filter(lambda nota: float(nota) < NOTA_MINIMA, lista_notas))

print("Notas ingresadas:", lista_notas)
print("Aprobados:", aprobados_user)
print("Reprobados:", reprobados_user)


# 3. FUNCIÓN REDUCE
# Reducción de una secuencia a un único valor acumulado

# Acumulación de la suma de solo los números pares
numeros_reduccion = [1, 2, 8, 4, 9, 6]
pares_filtrados = filter(lambda num: num % 2 == 0, numeros_reduccion)

suma_pares = reduce(lambda acumulador, num: acumulador + num, pares_filtrados, 0)

print("Números original:", numeros_reduccion)
print("Suma acumulada de pares:", suma_pares)
```

```python
# TUPLAS: BASES Y USO CON FECHAS

# 1. Ejemplo básico con tuplas
tupla = (5, 2, 9, 6, 8, 5)

print("Tupla completa:", tupla)
print("Primer valor:", tupla[0])
print("Último valor:", tupla[-1])
print("Cantidad de 5s:", tupla.count(5))
print("Rango (índices 2 al 4):", tupla[2:5])

# Recuerda: las tuplas no se pueden modificar tras crearse (son inmutables)


# 2. Uso de tuplas para formatear fechas
import datetime

fecha = datetime.date.today()
dia = fecha.day
mes = fecha.month
year = fecha.year

print("Fecha:", fecha, "| Día:", dia, "| Mes:", mes, "| Año:", year)

semana = ("domingo", "lunes", "martes", "miércoles", "jueves", "viernes", "sábado")
meses = ("", "Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre")

# Obtenemos el nombre del día actual
# Usamos % 7 para asegurar que el domingo (día 7) apunte al índice 0
dia_semana = semana[fecha.isoweekday() % 7]

print("Primer día de la tupla:", semana[0])
print(dia_semana, dia, "de", meses[mes], "del año", year)
```

```python
from functools import reduce

# Las listas pueden contener cualquier valor desde objetos hasta tipos de datos
lista = [5, 2, 9, 4, "Programacion Computacional III", [8, 9, 8], True]
print(lista)

# Ejercicio: haciendo uso de las listas en python calcule la nota final de computo I
notas = [
    # Computo I
    ["Lab1", 8, 0.3],
    ["Lab2", 9, 0.3],
    ["parcial1", 10, 0.4],
    # Computo II
    ["Lab1", 10, 0.3],
    ["Lab2", 9, 0.3],
    ["parcial1", 10, 0.4],
    # Computo III
    ["Lab1", 7, 0.3],
    ["Lab2", 8, 0.3],
    ["parcial1", 6, 0.4],
]

# Funcion encargada de calcular la nota
def calcularNf(nota, notas):
    return notas[1] * notas[2] + nota

# Calculo de la nota final acumulada
nf = reduce(calcularNf, notas, 0) / 3
print("NF: ", nf)
```

```python
# Estructuras de control

# if, while
# hacer un ejercicio en python que perita clasificar los rangos de edades,
# hasta que el usuario ingrese un valor incorrecto
seguir = True
while seguir:
    edad = int(input("Edad: "))

    if edad < 0:
        seguir = False
        print("Edad incorrecta")
    elif edad <= 2:
        print("Eres un bebe")
    elif edad <= 11:
        print("Eres un niño")
    elif edad < 18:
        print("Eres un adolescente")
    elif edad < 65:
        print("Eres un adulto")
    elif edad <= 85:
        print("Eres un adulto mayor")
    elif edad <= 130:
        print("Larga vida")
    else:
        print("Edad incorrecta")
        seguir = False
```
