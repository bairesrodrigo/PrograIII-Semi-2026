```python
from functools import reduce

# 1. FUNCIÓN MAP
# Transformación elemento a elemento

# A. Extracción de iniciales de días de la semana
semana = ["domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado"]
print(f"Semana: {semana} | Primer día: {semana[0]}")

primera_letra = list(map(lambda dia: dia[0], semana))
print("Primeras letras:", primera_letra)

primera_letra_mayus = list(map(lambda dia: dia[0].upper(), semana))
print("Primeras letras (mayúsculas):", primera_letra_mayus)

# B. Elevar lista a un exponente dinamico
numeros_base = [2, 4, 6, 7, 9]
exponente = int(input("Exponente: "))
potencias = list(map(lambda num: num ** exponente, numeros_base))
print("Resultados potencia:", potencias)

# C. Generar tabla de multiplicar
numeros_tabla = list(range(1, 11))
tabla = int(input("Tabla a multiplicar: "))
resultado_tabla = list(map(lambda num: num * tabla, numeros_tabla))
print(f"Tabla del {tabla}:", resultado_tabla)


# 2. FUNCIÓN FILTER
# Filtrado condicional de elemento

# A. Clasificación de números pares e impares
serie = [2, 5, 6, 8, 9, 10, 12]
pares = list(filter(lambda num: num % 2 == 0, serie))
impares = list(filter(lambda num: num % 2 != 0, serie))

print("Original:", serie)
print("Pares:", pares)
print("Impares:", impares)

# B. Clasificación de notas (Lista estática)
notas = [1, 6, 8, 2, 7, 10, 5]
NOTA_MINIMA = 6

aprobados = list(filter(lambda nota: nota >= NOTA_MINIMA, notas))
reprobados = list(filter(lambda nota: nota < NOTA_MINIMA, notas))

print("Notas:", notas)
print("Aprobados:", aprobados)
print("Reprobados:", reprobados)

# C. Clasificación de notas desde input de usuario
entrada_usuario = input("Notas separadas por coma (ej: 10,8.8,6.5,4.5,3): ")
lista_notas = entrada_usuario.split(",")

aprobados_user = list(filter(lambda nota: float(nota) >= NOTA_MINIMA, lista_notas))
reprobados_user = list(filter(lambda nota: float(nota) < NOTA_MINIMA, lista_notas))

print("Notas ingresadas:", lista_notas)
print("Aprobados:", aprobados_user)
print("Reprobados:", reprobados_user)


# 3. FUNCIÓN REDUCE
# Reducción de una secuencia a un único valor acumulado

# Acumulación de la suma de solo los números pares
numeros_reduccion = [1, 2, 8, 4, 9, 6]
pares_filtrados = filter(lambda num: num % 2 == 0, numeros_reduccion)

suma_pares = reduce(lambda acumulador, num: acumulador + num, pares_filtrados, 0)

print("Números original:", numeros_reduccion)
print("Suma acumulada de pares:", suma_pares)
```
